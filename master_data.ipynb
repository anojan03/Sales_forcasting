```python
invalid_dept = df[
    df['Department_ID'].isna() |
    (df['Department_ID'] == 0)
]

print("Total records:", len(df))
print("Records without valid Department ID:", len(invalid_dept))
print("Percentage:",
      round(len(invalid_dept) / len(df) * 100, 2), "%")
```

```python
invalid_dept[
    ['Transaction_ID',
     'Product_Name',
     'Department_ID',
     'Quantity',
     'Line_Total',
     'Promotion']
].head(30)
```

```python
df_clean = df[
    df['Department_ID'].notna() &
    (df['Department_ID'] != 0)
].copy()

df_clean.reset_index(drop=True, inplace=True)

print("Before cleaning:", len(df))
print("After cleaning:", len(df_clean))
print("Removed:", len(df) - len(df_clean))
```
